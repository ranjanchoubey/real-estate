# **Automatic EDA using Pandas Profiling**

- Pandas Profiling automates the process of generating a comprehensive report that provides insights into 
the structure and characteristics of a dataset. 

- The report includes statistics, visualizations, 
and correlations, making it easier to understand the data at a glance. 

### Suggestion :  Use Auto EDA and Must Do Manually for better understanding
---



```python
import pandas as pd
from ydata_profiling import ProfileReport

df = pd.read_csv('../data/clean-data/flats-house-cleaned-v2.csv')
profile = ProfileReport(df, title="YData Profiling Report", explorative=True)
profile.to_file("ydata_profiling_report.html")
```

```python
# view the report in the notebook 
profile.to_notebook_iframe()
```
